# Power och flerfaldig testning
När ett experiment planeras behöver vi hantera två olika risker:  
- att missa en effekt som faktiskt finns
- att hitta en effekt som egentligen bara beror på slumpen
Power handlar om den första risken. Flerfaldig testning handlar om den andra risken när många hypotesprövningar genomförs samtidigt.  
I den här lektionen fördjupar vi sambandet mellan signifikansnivå, power, effektstorlek och stickprovsstorlek. Vi jämför även Bonferroni och FDR som två olika sätt att begränsa falska positiva resultat.

#### Fyra möjliga resultat från ett hypotesprov
Ett hypotesprov kan ge fyra möjliga kombinationer av verklighet och beslut.

**1. Nollhypotesen är sann och vi förkastar den inte**  
Detta är ett korrekt beslut. Testet ger inte stöd för en effekt som inte finns.

**2. Nollhypotesen är sann men vi förkastar den**  
Detta är ett typ I-fel, även kallat ett falskt positivt resultat eller falsklarm.
Signifikansnivån alfa anger den högsta accepterade sannolikheten för ett typ I-fel i ett enskilt test, under förutsättning att testets antaganden gäller.
Om alfa är 0,05 accepterar vi en felrisk på 5% ör ett sådant test.

**3. En effekt finns men vi förkastar inte nollhypotesen**
Detta är ett typ II-fel, även kallat ett falskt negativt resultat.
Sannolikheten för ett typ II-fel betecknas med beta.

**4. En effekt finns och vi förkastar nollhypotesen**
Testet har då upptäckt effekten. Sannolikheten för detta är testets power:  
**power = 1 − beta**  
Om beta är 0,20 är powern 80%

### Vad betyder 80 procent power?
En power på 80 % betyder att experimentupplägget, upprepat många gånger under samma förutsättningar, skulle upptäcka en effekt av den planerade storleken i ungefär 80 % av experimenten där effekten faktiskt finns.  
Det betyder INTE att:  
- nollhypotesen har 80 % sannolikhet att vara falsk
- ett signifikant resultat har 80 % sannolikhet att vara korrekt
- testet upptäcker alla tänkbara effekter med 80 % sannolikhet  
Power gäller för en bestämd effektstorlek och ett bestämt experimentupplägg.

### Sambandet mellan alfa och power
Signifikansnivån och powern är inte oberoende av varandra.  
Om vi sänker alfa från 0,05 till 0,01 blir det svårare att förkasta nollhypotesen. Risken för typ I-fel minskar, men powern minskar normalt också om stickprovsstorleken och effektstorleken är oförändrade.  
För att behålla samma power med en lägre signifikansnivå behöver vi vanligtvis ett större stickprov.  
Det finns alltså en avvägning mellan:  
- risken för falska positiva resultat
- risken för falska negativa resultat
- experimentets storlek och kostnad  
Vilken avvägning som är rimlig beror på konsekvenserna av respektive fel.

### Minsta detekterbar effekt
När stickprovsstorleken planeras behöver vi ange vilken effekt experimentet ska ha en rimlig chans att upptäcka. Denna effekt kallas ofta **minsta detekterbar effekt**, eller **MDE**.
MDE ska helst motsvara den minsta skillnad som är praktiskt viktig.  
Om konverteringsgraden i kontrollgruppen är 25 % och vi vill kunna upptäcka en ökning till 27 % är den planerade skillnaden:  
**27 % − 25 % = 2 procentenheter**  
MDE betyder inte att mindre effekter är omöjliga att observera. Ett experiment kan råka ge ett signifikant resultat för en mindre effekt. Däremot är experimentet inte planerat för att på ett tillförlitligt sätt upptäcka sådana mindre effekter.  
Om vi vill kunna upptäcka en mindre effekt med samma power behöver vi normalt ett större stickprov.

### Vad påverkar stickprovsstorleken?
En powerberäkning kräver flera antaganden:  
- kontrollgruppens förväntade utgångsnivå
- minsta praktiskt betydelsefulla effekt
- önskad power
- signifikansnivå
- ensidigt eller tvåsidigt test
- fördelningen mellan grupperna
- vilken typ av utfallsvariabel som analyseras
- den förväntade variationen för kontinuerliga utfall  

Beräkningen är bara så bra som dessa antaganden. Om den verkliga konverteringsgraden eller variationen skiljer sig mycket från planeringen kan experimentets faktiska power också bli annorlunda.  
Lika stora grupper är ofta mest effektiva när observationerna kostar ungefär lika mycket i båda grupperna. En ojämn fördelning kan ändå vara motiverad om grupperna har olika kostnader eller risker.

### Effektstorlek i powerberäkningen
Statistikbibliotek använder ofta standardiserade effektstorlekar.  
För två andelar kan statsmodels använda Cohen’s h. Den omvandlar två andelar till ett standardiserat mått som powerfunktionen kan arbeta med.  
För två medelvärden används ofta **Cohen’s d:**  
**Cohen’s d = skillnaden mellan medelvärdena / standardavvikelsen**  
Den standardiserade effektstorleken behövs för beräkningen, men resultatet bör också uttryckas i en enhet som är begriplig i verksamheten, exempelvis:  
- procentenheter
- kronor
- minuter
- antal köp  
Ett standardiserat mått ersätter alltså inte den praktiska tolkningen.

### Analytisk och simulerad power
Power kan beräknas på två sätt.  

**Analytisk beräkning**  
En färdig matematisk metod används för att beräkna power eller nödvändig stickprovsstorlek utifrån angivna antaganden.  
Detta är vanligtvis snabbt och passar etablerade test med välkända formler.

**Simulering**
Experimentet simuleras många gånger. Power uppskattas som andelen simuleringar där testet förkastar nollhypotesen.  
Simulering är användbar för att:
- kontrollera en analytisk beräkning
- förstå hur power fungerar
- undersöka mer komplicerade experimentupplägg  
Den simulerade powern varierar något mellan körningar. Ju fler simuleringar som används, desto stabilare blir uppskattningen, men beräkningen tar också längre tid.
Den analytiska och simulerade powern bör ligga ungefär nära varandra när de bygger på samma antaganden.

### Problemet med många hypotesprövningar
Signifikansnivån 0,05 gäller ett enskilt test. Om många nollhypoteser testas ökar risken att minst ett resultat blir signifikant av en slump.  
Med 20 oberoende test där alla nollhypoteser är sanna förväntar vi oss i genomsnitt ungefär:  
**20 × 0,05 = 1 falskt positivt resultat**  
Detta är ett förväntat antal, inte en garanti att exakt ett test blir signifikant.
Sannolikheten för minst ett falskt positivt resultat är i detta exempel ännu högre än 5 %. För 20 oberoende test är den ungefär:
**1 − (1 − 0,05)²⁰ ≈ 64 %**


### Vad är en testfamilj?
En testfamilj är en grupp hypotesprövningar som hör samman och tillsammans används för en slutsats eller ett beslut.  
Exempel kan vara:  
- samma primära utfall i flera målgrupper
- flera varianter som jämförs med samma kontroll
- flera mätvariabler som används för att avgöra om en produkt ska lanseras

Alla test i ett helt projekt behöver inte automatiskt tillhöra samma familj. Indelningen bör bestämmas utifrån vilka test som tillsammans stödjer samma påstående eller beslut.  
Detta bör planeras innan resultaten undersöks.


### Bonferroni och familjevis felrisk
Den familjevisa felrisken, ofta förkortad FWER, är sannolikheten att göra minst ett typ I-fel inom testfamiljen.  
Bonferronimetoden begränsar denna risk genom att dela den önskade totala signifikansnivån med antalet test:  
justerad alfa = total alfa / antal test
Om 20 test genomförs och den totala nivån ska vara 0,05 blir gränsen:
0,05 / 20 = 0,0025  
Bonferroni är enkel och ger ett starkt skydd mot falska positiva resultat. Nackdelen är att metoden kan bli mycket strikt när många test genomförs. Då minskar powern och fler verkliga effekter kan missas.  
Metoden passar bäst när även ett enda falskt positivt resultat kan få stora konsekvenser.

### FDR och Benjamini-Hochberg
**False discovery rate**, **FDR**, är den förväntade andelen falska positiva fynd bland de hypoteser som förkastas.  
Benjamini-Hochberg är en vanlig metod för att kontrollera FDR. Metoden rangordnar p-värdena och jämför dem med gränser som beror på deras placering och antalet test.  
FDR är vanligtvis mindre strikt än Bonferroni. Det innebär att:  
- fler verkliga effekter kan upptäckas
- fler hypoteser kan förkastas
- en begränsad andel falska fynd accepteras bland de rapporterade fynden  
FDR passar ofta bättre i utforskande analyser där vi söker efter intressanta signaler som senare kan undersökas vidare.

### Välj metod efter konsekvensen
Bonferroni och FDR skyddar mot olika typer av risk.  
Bonferroni är ofta lämpligt när:  
- analysen är bekräftande
- antalet primära hypoteser är begränsat
- kostnaden för ett enda falskt positivt resultat är hög
**FDR är ofta lämpligt när:**  
- analysen är utforskande
- många hypoteser undersöks
- målet är att hitta kandidater för fortsatt analys
- en mindre andel falska fynd kan accepteras  
Valet ska göras utifrån analysens syfte och helst bestämmas innan resultaten granskas.  
När korrigerade p-värden används kan de jämföras med den ursprungliga signifikansnivån. Alternativt kan råa p-värden jämföras med en justerad gräns. Dessa två arbetssätt ska inte blandas.

### En genomtänkt analysplan
En analysplan för ett experiment med flera hypoteser bör ange:  
- primärt utfall
- sekundära och utforskande utfall
- minsta praktiskt betydelsefulla effekt
- planerad stickprovsstorlek
- önskad power
- signifikansnivå
- vilka test som ingår i samma familj
- vilken korrigeringsmetod som ska användas
- när experimentet ska avslutas  
Genom att bestämma detta i förväg blir det lättare att skilja planerade slutsatser från intressanta fynd som behöver bekräftas i en ny analys.  

# Power och flerfaldig testning

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize
from statsmodels.stats.proportion import proportions_ztest
from statsmodels.stats.multitest import multipletests

np.random.seed(42)
```

```python
baseline_rate_A = 0.25
min_detectable_diff_pp = 2.0
min_detectable_diff = min_detectable_diff_pp / 100.0

target_rate_B = baseline_rate_A + min_detectable_diff
alpha = 0.05
power_target = 0.80
```

```python
effect_size = proportion_effectsize(prop1=baseline_rate_A, prop2=target_rate_B)

solver = NormalIndPower()
n_per_group = solver.solve_power(effect_size=effect_size, 
                                 alpha=alpha, 
                                 power=power_target, 
                                 ratio=1.0,
                                 alternative='two-sided')

n_per_group_rounded = int(np.ceil(n_per_group))
effect_size, n_per_group, n_per_group_rounded
```
