# Lesson 5
#### File reading


```python
file = open("files/example.txt", "r") # r = read
content = file.read()   # reads the whole file as a string
file.close()  # SUPER IMPORTANT to close the file afterwards

print(content)
```

#### File reading in a better way

```python
with open("files/example.txt") as file:   # better way, closes the file automatically
    file_content = file.read()

print(file_content)
```

#### File reading line by line

```python
with open("files/example.txt") as file:
    first_line = file.readline()
    second_line = file.readline()

    file.seek(0)  # sets the line to read to the index line in the file 0 to set it from start

print(first_line)
print(second_line)
```

```python
with open("files/example.txt") as file:
    lines = file.readlines()  # get a list of all lines as strings, separated with a comma

for line in lines:
    print(line)
```

#### File reading line by line with loop printing

```python
with open("files/example.txt",encoding="utf-8") as file:   # always use encoding, utf8 takes åäö. emojis, japanese mm
    for line in file:          # loops the print directly, strip removes the \n and extra spaces
        print(line.strip())
```

## Error handling
#### try/except


```python
try:
    with open("files/missing.txt", "r", encoding="utf-8") as f:
        data = f.read()

    print(data)
except Exception as e:
    print(f"Något gick fel. Error code: {e}")
```

#### File reading with loop printing with enumerate for numbering lines


```python
with open("files/example.txt", "r", encoding="utf-8") as file:
    for i, line in enumerate(file, start=1):
        print(f"{i}: {line.strip()}")
```

#### File reading, ignoring empty lines and spacces.  Converting to numbers with mathematical operations 

```python
values = []
with open("files/numbers.txt", "r", encoding="utf-8") as numbers_file:
    for line in numbers_file:
        line = line.strip()
        if line:  # ignores empty lines
            values.append(float(line))

if values:
    avg = sum(values) / len(values)
    print(f"Medelvärde: {avg}")
else: 
    print("Inga värden i listan")
```

# File writing

```python
with open("files/example2.txt", "w", encoding="utf-8") as f: # w = write
    f.write("Hello world\n")
    f.write("Det här andra raden i filen\n")
```

#### File writing with append, to avoid writing over information in a file

```python
with open("files/example2.txt", "a", encoding="utf-8") as f: # a = append
    f.write("Detta ska bli rad 3\n")
```

#### File writing with writelines

```python
lines = ["Rad 1\n", "Rad 2\n", "Rad 3\n"]
with open("files/example2.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)    # no need for a loop, writes a iterable like a a list line by line to file
```

#### File writing with print()

```python
with open("files/example2.txt", "w", encoding="utf-8") as f:
    print("Hello world", file=f)
```

#### File writing example with input loop

```python
while True:
    note = input("Skriv en anteckning (eller q för att avsluta):")
    if note.lower().strip() == "q":
        break
    with open("files/example2.txt", "a", encoding="utf-8") as file:
        file.write(note + "\n")
```

# CSV & JSON files
- CSV = Comma Separated Values
- JSON = JavaScript Object Notation

#### CSV Writer

```python
import csv

with open("files/cars.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Reg", "Color"])
    writer.writerow(["ABC123", "Red"])
    writer.writerow(["ZYX789", "Blue"])
```

#### CSV Reader

```python
with open("files/cars.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
        print(row[1]) # print just the color
        print(row[0]) # print just the reg nr
```

## JSON
#### Writing to file

```python
import json

data = {
    "name": "Håkan",
    "age": "47",
    "city": "malmö"
}

with open("files/person.json", "w", encoding="utf-8") as file:
    json.dump(data, file, indent=4, ensure_ascii=False)
```

#### Reading from file

```python
with open("files/person.json", "r", encoding="utf-8") as file:
    person = json.load(file)

print(person)
print(person["name"])
```
