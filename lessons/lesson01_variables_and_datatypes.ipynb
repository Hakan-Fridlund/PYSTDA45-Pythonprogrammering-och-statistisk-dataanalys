## Numbers

#### Operators

```python
print(7 // 3)  # whole number division
print(7 % 3)   # modulus operator (remainder)
print(7 ** 3)  # exponentiation
```

#### Math functions

```python
num = -7

print(abs(num))  # absolute value
print(round(3.14159, 2))  # round to 2 decimal places
print(round(3.14159))  # round to nearest integer
print(pow(2, 3))  # 2 raised to the power of 3
```

#### Import math library for extra functions

```python
import math

print(math.sqrt(16))  # square root
print(math.pi)  # value of pi
print(math.e)  # value of e
print(math.ceil(3.2))  # round up
print(math.floor(3.8))  # round down
```

## Strings  
\n # new line  (\ escape sign in a string for a command)  

```python
name = "Håkan F"
print(f"Hello my name is {name}!")  # f-string
print(len(name))  # length of string
print(name[1])  # indexing the string (second character)
print(name[-1])  # indexing the string (last character)
print(name[0:4])  # slicing the string (first 4 characters)

text = "        hello world!      "
print(text.strip().title())  #strip removes whitespace from both ends,
                             #title capitalizes the first letter of each word

words = text.strip().split()  # split the string into a list of words
                              # .split(",") would split on commas instead of whitespace
print(words[0]) # print the first word
print(type(words)) # print the type of the variable words
```
