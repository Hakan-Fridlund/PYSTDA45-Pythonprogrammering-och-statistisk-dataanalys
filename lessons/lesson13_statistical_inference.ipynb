# Statistisk inferens
Statistisk inferens innebär att vi använder information från ett stickprov för att dra slutsatser om en större population. Eftersom stickprovet bara innehåller en del av populationen behöver vi samtidigt beskriva hur osäkra slutsatserna är.

Från population till beslut
Ett förenklat statistiskt arbetsflöde kan beskrivas så här:
- Population → stickprov → skattning → osäkerhet → beslut
- Populationen är hela den grupp vi vill uttala oss om.
- En parameter är ett värde som beskriver populationen, exempelvis populationens verkliga medelvärde.
- Ett stickprov är de observationer vi faktiskt har samlat in.
- En skattning är ett värde som beräknas från stickprovet och används för att uppskatta parametern.

Stickprovsmedelvärdet kan exempelvis användas som en skattning av populationens medelvärde. Ett annat stickprov hade sannolikt gett ett något annorlunda resultat. Därför behöver skattningen kompletteras med ett mått på osäkerheten.
Statistiska metoder kan inte kompensera för ett dåligt urval eller en olämplig studiedesign. Ett stort men snedvridet stickprov kan fortfarande ge en missvisande bild av populationen.

#### Tre delar i en statistisk slutsats
En statistisk analys bör vanligtvis besvara tre olika frågor.
**Hur stor är den observerade effekten?**
Effektskattningen beskriver skillnadens storlek och riktning. Det kan exempelvis vara:
- en skillnad mellan två medelvärden
- en skillnad mellan två andelar
- ett samband mellan två variabler

#### Hur osäker är skattningen?
Ett konfidensintervall visar ett intervall av värden som är förenliga med data och den använda statistiska modellen.
- Ett brett intervall visar stor osäkerhet.  
- Ett smalt intervall visar att skattningen är mer precis.

### Hur ovanligt är resultatet under nollhypotesen?
Ett p-värde beskriver hur ovanligt det observerade resultatet, eller ett mer extremt resultat, skulle vara om nollhypotesen och testets övriga antaganden gäller.
P-värdet berättar inte hur stor eller praktiskt viktig effekten är. Därför bör vi normalt redovisa både effektskattningen, konfidensintervallet och p-värdet.

##### Medelvärde eller andel?
Vilken jämförelse vi gör beror på vilken typ av utfallsvariabel vi har. 

#### Kontinuerligt utfall
Ett kontinuerligt utfall kan anta många numeriska värden, exempelvis:
- köpesumma
- leveranstid
- blodtryck
- tid på en webbsida
För sådana utfall jämför vi ofta gruppernas medelvärden. För två oberoende grupper kan exempelvis ett Welch-test användas.
95 % CI för skillnad i medel.  
**Rapportera alltid:**  
- Skattad skillnad i medel
- 95 % CI
- p-värde
- effektens storlek i ord

#### Binärt utfall
Ett binärt utfall har två möjliga värden, exempelvis:
- klickade eller klickade inte
- köpte eller köpte inte
- sjuk eller inte sjuk  
För sådana utfall jämför vi vanligtvis gruppernas andelar.  
Om 20 av 100 personer genomför ett köp är andelen 20 / 100 = 0,20 = 20%  
Z-test för två andeler, 95 CI för skillnaden i andelar.
**Rapportera alltid:**
- Skattad skillnad i andel (pp)
- 95 CI
- p-värde
- effektens storlek i ord

#### Procent och procentenheter
Procent och procentenheter beskriver olika saker.
Anta att konverteringsgraden ökar från 20 % till 25 %.
Den absoluta skillnaden är:
25 % − 20 % = 5 procentenheter
Den relativa ökningen är:
(25 % − 20 %) / 20 % = 25 %
Resultatet kan alltså beskrivas som en ökning med 5 procentenheter eller en relativ ökning med 25 procent. I en rapport behöver det framgå vilket av dessa mått som används.
När två andelar jämförs i ett A/B-test är skillnaden i procentenheter ofta den tydligaste effektskattningen.

### Tvåsidigt och ensidigt test
Ett tvåsidigt test undersöker om det finns en skillnad i någon riktning.
- Är grupp A bättre än grupp B?
- Är grupp A sämre än grupp B?

Ett ensidigt test undersöker endast en förutbestämd riktning, exempelvis om grupp B har en högre konverteringsgrad än grupp A.
Ett tvåsidigt test är vanligtvis ett lämpligt standardval. Ett ensidigt test bör endast användas när riktningen har bestämts innan data analyseras och den motsatta riktningen inte är relevant för frågeställningen.
Vi får inte först undersöka resultatet och därefter välja den riktning som ger det minsta p-värdet.

### Statistisk och praktisk betydelse
Ett statistiskt signifikant resultat behöver inte vara praktiskt viktigt.
Med ett mycket stort stickprov kan även en liten skillnad bli statistiskt signifikant. En ökning av konverteringsgraden med 0,1 procentenheter kan exempelvis vara för liten för att motivera kostnaden för en förändring.
Innan analysen bör vi därför fundera på vilken effekt som är tillräckligt stor för att ha praktisk betydelse. Detta kan kallas den **minsta praktiskt betydelsefulla effekten.**
Ett beslut bör baseras på:
- effektens storlek och riktning
- konfidensintervallets bredd
- resultatet från hypotesprövningen
- kostnader, risker och verksamhetens sammanhang

## Power och två typer av fel
När vi genomför ett hypotesprov finns två viktiga felrisker.
- Ett **falskt positivt resultat**, eller typ I-fel, innebär att vi förkastar en sann nollhypotes.
- Ett **falskt negativt resultat**, eller typ II-fel, innebär att vi inte upptäcker en verklig effekt.

**Statistisk power** är sannolikheten att testet upptäcker en effekt av en viss storlek när effekten faktiskt finns.
Power påverkas bland annat av:
- stickprovets storlek
- effektens storlek
- variationen i data
- den valda signifikansnivån

Ett litet stickprov kan ha låg power. Att resultatet inte är statistiskt signifikant betyder då inte nödvändigtvis att det saknas en relevant effekt.

#### Problemet med många tester
Om vi genomför många hypotesprövningar ökar sannolikheten att minst ett resultat blir statistiskt signifikant av en slump.
Vid signifikansnivån 0,05 förväntas ungefär ett av tjugo oberoende test av sanna nollhypoteser bli signifikant enbart genom slumpvariation.
Därför bör vi före analysen:
- bestämma vilket utfall som är viktigast
- begränsa antalet tester
- skilja mellan planerade och utforskande analyser
- vid behov använda en metod som justerar för flerfaldig testning

#### Innan ett A/B-test genomförs
För att minska risken för missvisande slutsatser bör följande bestämmas innan resultatet undersöks:
- vilken primär mätvariabel som ska analyseras
- vilken population undersökningen gäller
- hur deltagarna fördelas mellan grupperna
- om testet ska vara tvåsidigt eller ensidigt
- vilken skillnad som är praktiskt betydelsefull
- vilken signifikansnivå som ska användas
- hur stort stickprovet behöver vara
- när datainsamlingen ska avslutas

Vid ett experiment behöver fördelningen mellan grupperna vara slumpmässig för att en observerad skillnad på ett trovärdigt sätt ska kunna kopplas till den förändring som testas.
