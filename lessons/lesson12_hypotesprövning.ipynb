# Hypotesprövning

### Tre frågor i en statistisk jämförelse
När två grupper jämförs behöver vi vanligtvis besvara tre olika frågor:
- 1. Hur stor skillnad ser vi? 
Det beskriver vi exempelvis med skillnaden mellan gruppernas medelvärden.

- 2. Hur osäker är uppskattningen?
Det kan beskrivas med ett konfidensintervall.

- 3. Hur ovanlig vore skillnaden om grupperna egentligen inte skiljde sig?
Det undersöker vi med ett hypotesprov och ett p-värde.

Dessa frågor kompletterar varandra. Ett p-värde berättar inte hur stor eller praktiskt viktig en skillnad är.

## Nollhypotes och alternativhypotes
Antag att vi jämför populationsmedelvärdet i grupp A och grupp B.  
Nollhypotesen, H₀:  
μB − μA = 0  
Det innebär att det inte finns någon skillnad mellan populationernas medelvärden.  
De skillnader vi ser i stickprovet förklaras då av slumpmässig variation.    
Alternativhypotesen, H₁:  
μB − μA ≠ 0  
Detta är en tvåsidig alternativhypotes. Den undersöker om grupperna skiljer sig, oavsett vilken grupp som har det högsta medelvärdet.  
Hypoteserna bör formuleras innan resultatet analyseras.

### P-värdet
P-värdet besvarar följande fråga:  
Om nollhypotesen och testets antaganden gäller, hur vanligt är det då att få ett resultat som är minst lika extremt som det vi observerade?
Ett litet p-värde innebär att resultatet är ovanligt under nollhypotesen och ger evidens mot den.
P-värdet är däremot INTE:
- sannolikheten att nollhypotesen är sann
- sannolikheten att resultatet beror på slumpen
- ett mått på hur stor eller viktig effekten är

### Signifikansnivå och beslut
Innan testet väljer vi en signifikansnivå, α. Ett vanligt val är 0,05.  
Om p < α förkastar vi nollhypotesen.  
Om p ≥ α kan vi inte förkasta nollhypotesen.  
Att inte kunna förkasta nollhypotesen är inte samma sak som att bevisa att grupperna är identiska. Resultatet kan även bero på stor osäkerhet eller för lite data.  
Gränsen 0,05 är en överenskommen beslutsgräns, inte en skarp gräns mellan sant och falskt.

### Teststatistikan
Ett hypotesprov jämför den observerade effekten med hur mycket resultatet förväntas variera av slumpen.  
När nollhypotesen säger att skillnaden är noll kan grundidén skrivas:  
**teststatistika = observerad skillnad / skillnadens standardfel**  
Teststatistikan visar ungefär hur många standardfel den observerade skillnaden ligger från noll.  
- Tecknet visar skillnadens riktning och beror på i vilken ordning grupperna jämförs.
- Absolutvärdet visar hur långt resultatet ligger från noll i förhållande till osäkerheten.
- Ett stort absolutvärde ger normalt ett mindre p-värde eftersom resultatet är mer ovanligt under nollhypotesen.
I ett t-test kallas denna test för ett t-värde.

### T-test och Welch-test
Ett t-test kan användas för att jämföra medelvärdena i två oberoende grupper.  
Med oberoende grupper menas att varje observation tillhör en grupp och att en observation i den ena gruppen inte hör ihop med en bestämd observation i den andra. Om samma personer exempelvis mäts före och efter en behandling är observationerna i stället parade och då behövs en annan typ av test.  
Det klassiska t-testet antar att populationerna har samma varians. Welch-testet kräver inte detta antagande och är därför ofta ett lämpligt standardval när två oberoende gruppmedelvärden jämförs.  
Testet bygger fortfarande på att observationerna är tillräckligt oberoende och att datainsamlingen är rimlig för den slutsats vi vill dra. Fördelningar och möjliga extremvärden bör därför undersökas innan resultatet tolkas.

### Permutationstest
Ett permutationstest skapar en nollfördelning genom att simulera situationen där grupptillhörigheten inte har någon betydelse.
Ett förenklat arbetsflöde är:  
- 1. Beräkna den observerade skillnaden mellan grupperna.
- 2. Lägg ihop gruppernas värden.
- 3. Blanda grupptillhörigheten slumpmässigt, men behåll gruppernas storlekar.
- 4. Beräkna en ny skillnad.
- 5. Upprepa blandningen många gånger.
- 6. Beräkna hur stor andel av skillnaderna som är minst lika extrema som den observerade.

Den andelen blir ett uppskattat p-värde. Metoden bygger på att värdena, om nollhypotesen gäller, skulle kunna byta gruppetikett utan att vi bryter mot hur datan samlades in. För parade eller beroende observationer måste permutationstestet anpassas till studiens upplägg.

### Bootstrap och permutation har olika uppgifter
När vi resamplar med återläggning inom de ursprungliga grupperna får vi en bootstrapfördelning som vanligtvis är centrerad kring den observerade skillnaden.  
Den passar bra för att uppskatta effektens osäkerhet och skapa ett konfidensintervall.  
Den är däremot inte automatiskt en nollfördelning. För ett p-värde behöver vi simulera vad som händer när nollhypotesen gäller, exempelvis med:  
- ett permutationstest där gruppetiketterna blandas
- en bootstrap där grupperna först centreras till samma medelvärde
Kort sammanfattat:  
- Bootstrap inom grupperna: Hur osäker är den uppskattade skillnaden?
- Permutationstest: Hur ovanlig är skillnaden om det inte finns någon gruppeffekt?

### Sambandet med konfidensintervallet
Ett konfidensintervall och ett tvåsidigt hypotesprov beskriver samma data från olika perspektiv.  
När samma metod och antaganden används gäller normalt följande:  
- Om ett 95-procentigt konfidensintervall för skillnaden inte innehåller noll blir det tvåsidiga p-värdet mindre än 0,05.
- Om intervallet innehåller noll kan vi inte förkasta nollhypotesen på signifikansnivån 0,05.
Konfidensintervallet ger dessutom information om vilka effektstorlekar som är rimliga. Därför bör både intervallet och p-värdet redovisas.

### Effektstorlek och praktisk betydelse
Statistisk signifikans betyder inte automatiskt att skillnaden är praktiskt viktig.
Cohen’s d beskriver skillnaden mellan två medelvärden i förhållande till gruppernas typiska spridning. Ett större absolutvärde betyder en större standardiserad skillnad.  
Riktvärden som 0,2, 0,5 och 0,8 används ibland för små, medelstora och stora effekter, men de måste alltid tolkas i sitt sammanhang.
En tydlig redovisning bör därför innehålla:  
- den observerade skillnaden
- ett konfidensintervall
- vilket test som användes och dess p-värde
- en relevant effektstorlek
- en tolkning i analysens verkliga sammanhang

creates a synthetic dataframe with random nan-values and duplicates

```python
np.random.seed(42)
n_total = 200
n_A = n_total // 2
n_B = n_total - n_A

dates = pd.date_range('2026-01-01', periods=n_total, freq='D')

A_metric = np.random.normal(loc=50, scale=10, size=n_A)
B_metric = np.random.normal(loc=53, scale=11, size=n_B)

df = pd.DataFrame({
    'group':['A']*n_A + ['B']*n_B,
    'metric': np.concatenate([A_metric, B_metric]),
    'date': dates
})

nan_idx = np.random.choice(df.index, size=5, replace=False)
df.loc[nan_idx, 'metric'] = np.nan

dup_row = df.sample(1, random_state=17)
df = pd.concat([df, dup_row], ignore_index=True)

outliers_idx_A = df.query('group == "A" and metric.notna()', engine='python').sample(1, random_state=18).index
outliers_idx_B = df.query('group == "B" and metric.notna()', engine='python').sample(1, random_state=19).index
df.loc[outliers_idx_A, 'metric'] = df.loc[outliers_idx_A, 'metric'] +80
df.loc[outliers_idx_B, 'metric'] = df.loc[outliers_idx_B, 'metric'] -60

df.to_csv('ab_data.csv', index=False)
```
