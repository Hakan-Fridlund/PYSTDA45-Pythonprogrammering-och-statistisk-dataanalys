# Descriptive statistics
#### Datatypes
#### Categorical - labels, color, cities
- Nominal - no order, like brands, Volvo, Audio, Porsche et.c.
- Ordinal - natural order, like high, medium, low  
Can be used with shares, proportions. **Bar charts**, **pie charts**  
modes (most common) 

#### Numerical or quantitative data - numbers
- Discreet - whole numbers, int ex. number of purchases (no decimals)
- Continuous - hours, centimeters, weight  
median, mean,  std deviations  
**Bar chart** and **dot plot** for discreet  
**Histogram**, **box plot**, **line graph** and **density plot** for continuous


## Population and selection
**Statistical inference** is the process of using data collected from a smaller sample (**selection**) to draw conclusions and make predictions about the entire population.

#### Quartiles and interquartile range IQR
Q1, Q2, Q3 ,Q4  ex: 2,3,4 -- 5,5,6 -- 7,8,9 -- 10,12,14  
Q1 = 4,5 median of 4 and 5 --  Q3 = 9,5 median of 9 and 10
#### **IQR** = Q3-Q1 



## Variance and standard deviation
Z-points = (value - mean(avg)) / standard deviation  
positive Z means over average, negative Z means below average


### Outliers / extreme values
Q1 - (1,5 * IQR)  
Q3 + (1,5 * IQR)

# NumPy and numeric arrays

```python
import numpy as np

py_list = [1, 2, 3]
np_array = np.array([1, 2, 3])
print(py_list, np_array)
print(type(py_list))
print(type(np_array))
```

```python
print(py_list + [4, 5])
print(np_array + 10)
np_array = np_array.astype(np.float32) # changes the datatype
print (np_array.dtype)
```

```python
v = np.array([10, 20, 30, 40]) # 1D
m = np.array([[1, 2, 3],
              [4, 5, 6]])  # 2D --  2 rows, 3 columns
print(v.shape, m.shape)
```

```python
# arange creates an array with 0-11, reshape makes it into 2d (rows, columns)
np.arange(12).reshape(3,4)
```

```python
print(np.zeros((2, 3)))
np.linspace(0,1,5) # creates array with (startvalue, final value, steps)
```

```python
x = np.array([10, 20, 30, 40])
y = np.array([1, 2, 3, 4])
print(x * y)
print(x.mean()) # mean (avg), min(), max()
print(x.std()) # standard deviation
```

#### Broadcasting

```python
matrix = np.array([[1, 2, 3],
             [4, 5, 6]]) # shape (2, 3)
vector = np.array([10, 20, 30])

matrix_plus_vector = matrix + vector
print(matrix.shape, vector.shape, matrix_plus_vector)
print(matrix + 100) # 100 gets added to all elements. that is broadcasting
```

```python
z = np.arange(12).reshape(3, 4)
print(z)
print(z[1,2]) # row 1 = 2 since 0 is first index and col 2 = 3
print(z[:, 0]) # the first column
print(z[:, 1:3]) # columns 1 and 2, 3 is excluding
```

## Masks and boolean filters

```python
mask = z % 2 == 0
print(mask)
print(z[mask])
```

```python
scores = np.array([50, 60, 70, 80, 90], dtype=np.float64)
# counting a z-value
z = (scores - scores.mean() / scores.std())
# round(2) writes the value with 2 decimals
print(f"mean: {scores.mean()}, standard deviation: {scores.std()}, Z-points: {z.round(2)}")
```
