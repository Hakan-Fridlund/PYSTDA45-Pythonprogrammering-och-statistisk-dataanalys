### Uppskattning och osäkerhet
En punktuppskattning, exempelvis ett medelvärde, är inte hela resultatet. Ett konfidensintervall visar hur osäker uppskattningen är.  
I videon beräknas konfidensintervallet för den genomsnittliga dricksandelen på två sätt:  
- med normalapproximation och standardfel
- med bootstrap  

Bootstrap innebär att nya stickprov skapas genom att dra observationer med återläggning från det ursprungliga stickprovet.  
Bootstrap kräver färre antaganden om fördelningens form, men metoden är inte antagandefri. Det ursprungliga stickprovet behöver fortfarande vara relevant för populationen, och observationerna behöver ha samlats in på ett sätt som gör resamplingen rimlig.  
Att två metoder ger liknande intervall är positivt, men bevisar inte i sig att resultatet är korrekt.  

### Ensidiga och tvåsidiga test
Ett ensidigt test används när frågan har en bestämd riktning, exempelvis om ett medelvärde är högre i en viss grupp.  
Riktningen ska bestämmas innan resultatet undersöks. Om frågan bara är om grupperna skiljer sig, utan en förutbestämd riktning, används normalt ett tvåsidigt test.  
Ett stort p-värde innebär att datan inte ger tillräckligt stöd för alternativhypotesen. Det bevisar inte att grupperna är identiska.  
Därför bör resultatet redovisas med:  
- gruppstorlekar
- gruppers uppskattade värden
- den observerade skillnaden
- ett konfidensintervall när det är möjligt
- testmetod och p-värde
- en slutsats formulerad med rätt grad av säkerhet

### Regression som en del av analysen
Regressionen i videon undersöker hur dricks i dollar samvarierar med notans storlek.  
Lutningen beskriver den genomsnittliga förändringen i dricks när notan ökar. Den visar inte att en större nota i sig orsakar högre dricks.  
Residualdiagrammet används för att kontrollera om den linjära modellen verkar rimlig. Ett oregelbundet moln runt noll är önskvärt. Tydlig kurvatur, förändrad spridning eller avvikande punkter kan visa att modellen inte fångar hela sambandet.  
Modellen är en enkel baslinje. Den är inte automatiskt en färdig prognosmodell för nya restauranger, tidsperioder eller kundgrupper.  

### Tolkning och begränsningar
Datasetet innehåller 244 restaurangnotor från ett visst sammanhang. Resultaten beskriver i första hand detta dataset och ska inte utan vidare generaliseras till alla restaurangbesök.  
Det är observationsdata. Andra variabler kan påverka de samband som observeras. Måltid, veckodag, sällskapsstorlek och andra faktorer kan exempelvis hänga ihop. Analysen ger därför inte stöd för säkra slutsatser om orsak och verkan.  
Ett icke-signifikant resultat bör formuleras som att analysen inte fann tydligt stöd för en skillnad. Det ska inte formuleras som att grupperna har bevisats vara lika.
