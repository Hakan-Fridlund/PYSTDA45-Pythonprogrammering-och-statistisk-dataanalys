## Functions

```python
def greet(parameter = "Janne"):
    print(f"Hej! {parameter}, välkommen!")

greet("Argument") # calling the function with an argument
greet()
```

```python
def area(radius):
    return 3.14 * radius ** 2

def vol(area, height):
    return area * height

radius = int(input("Ange radien: "))

height = int(input("Ange höjden: "))
calc_vol = vol(area(radius), height) # first argument is a function call to another function

print(f"Volymen är: {calc_vol}")
```

## Modules, packages and libraries
-A **module** is a file that collects classes, functions and variables and can be imported.   
-A **package** is a folder with related modules that is imported with a common name   
##### create \_\_init\_\_.py (empty file) in a folder to make it a package
-A **library** is a larger collection of code (consisting of modules and packages) designed for one specific area or purpose. External libraries are typically installed using pip install, while built-in ones belong to Python's Standard Library.

```python
from statistics import mean
import math
import numpy as np

print(math.sqrt(16))
print(np.sqrt(16))
print(mean([1, 2, 3, 4, 5]))
```
