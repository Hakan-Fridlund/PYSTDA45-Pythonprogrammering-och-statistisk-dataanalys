# Cleaning and summarizing data with Pandas

```python
df2 = pd.read_csv("files/studietimmar.csv")

df2.columns = (df2.columns
              .str.strip()
              .str.replace(" ", "_")
              .str.lower()
              )
print(df2.head())
print()
print(df2.info())
print()
print(df2.isna().sum())
print()
print(df2.dtypes)
```

#### Cleaning NA and duplicates 

```python
df_clean = df2.dropna(subset=["provpoäng"]) # drops na values from column
df_clean.loc[df_clean["studietimmar"].isna(), "studietimmar"] = 0 # filter column and na values and sets them to 0

print(df_clean.isna().sum())
print("Dublettrader:", df_clean.duplicated().sum()) # shows duplicates
df_clean = df_clean.drop_duplicates().reset_index(drop=True)   # drops duplicates
print("Unika Studenter:", df_clean["student"].nunique())  # shows number of unique students
```

#### **Category** datatype in column

```python
# sets the column variable to category, this is done to save memory, if common values like A,B,C and D
df_clean["klass"] = df_clean["klass"].astype("category")
print(df_clean.dtypes)
```

#### make a new column while avoiding division by 0 with .where()

```python
# makes a new column, takes where provpoäng where studietimmar > 0 (to avoid division by 0) and divides by studietimmar
# if 0 it will be set to NA
df_clean = df_clean.assign(
    poang_per_timme = df_clean["provpoäng"]
    .where(df_clean["studietimmar"] > 0, other=pd.NA) 
    / df_clean["studietimmar"]
)
print(df_clean.head())
```

#### groupby summary

```python
summary_klass = (df_clean
                 .groupby("klass",dropna=False)
                 .agg(
                     antal = ("student", "count"),
                      medel_poäng = ("provpoäng", "mean"),
                      median_poäng = ("provpoäng", "median"),
                      std_poäng = ("provpoäng", "std"),
                      medel_timmar = ("studietimmar", "mean")
                      ).reset_index()
                 )
print(summary_klass)
```

#### crosstab with intervals from bins

```python
# bins sets the intervals, normalize="index" makes it into proportion % by row (alternative to index is columns and all)
# right = True sets the right number in the interval to inclusive
bins = [0, 60, 80, 100]
proportions = pd.crosstab(
    df_clean["klass"],
    pd.cut(df_clean["provpoäng"], bins=bins, right=True),
    normalize="index"
).round(2)

print(proportions)
```

#### **assert** debugging, if values is in interval otherwise error is raised and calculating correlation

```python
assert df_clean["provpoäng"].between(0, 100).all()
assert (df_clean["studietimmar"] >= 0).all()

# .corr() counts correlation value between to columns
corr = df_clean[["studietimmar", "provpoäng"]].corr().iloc[0, 1]
print("korrelation timmar-poäng", round(float(corr), 2))
```
