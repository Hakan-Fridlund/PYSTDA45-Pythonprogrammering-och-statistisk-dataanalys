## Indexing, changing and removal of elements in a list

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0]) # prints the first element/index of the list
print(fruits[-1]) # prints  the last element/index of the list
print(fruits[1:3]) # prints the second and third elements/indexes of the list

fruits[0] = "kiwi" # changes the first element/index of the list
fruits.append("orange") # adds a new element to the end of the list
print(fruits)

fruits.pop() # removes the last element of the list
fruits.remove("banana") # removes the first occurrence of the specified value
del fruits[0] # removes the first element of the list
print(fruits)
```

## Counting with elements in a list

```python
import statistics as stats
import numpy as np

numbers = [2, 2, 3, 7, 5]

print(len(numbers)) # prints the number of elements in the list
print(sum(numbers)) # prints the sum of the elements in the list
print(max(numbers)) # prints the maximum value in the list
print(numbers.count(2)) # prints the number of occurrences of the specified value in the list
print(sorted(numbers)) # prints a new list with the elements sorted in ascending order
numbers.sort() # sorts the list in place in ascending order

print(stats.mean(numbers)) # prints the mean of the list
print(stats.median(numbers)) # prints the median of the list

print(np.mean(numbers)) # prints the mean of the list using numpy
```

## Matrix / nested lists

```python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
matrix = [names, ages] # creates a 2D list (matrix) with names and ages

print (matrix[0][1]) # prints the second element of the first list (Bob)
print (matrix[1][2]) # prints the third element of the second list (30)
```

# Loops
### For loops
keywords: break and continue

```python
for i in range(0, 20, 4): # prints numbers from 0 to 16 with a step of 4
    print(i)

for name in range(len(names)): # prints the names in the list
    print(names[name])
```

```python
for i in range(10):
    if i == 5:
        break # exits the loop when i is equal to 5
    print(i) # prints numbers from 0 to 4
```

```python
for i in range(10):
    if i < 5:
        continue # skips the rest of the code in the loop when i is less than 5
    print(i) # prints numbers from 5 to 9
```

## Assignment:
create a loop from 0-> 15 where all even numbers is printed and count the even numbers.

```python
count = 0

for i in range(15):
    if i % 2 == 1:    # takes the uneven numbers and skips with continue
        continue
    print(i)
    count += 1
print(f"Count: {count}")
```

### While loops

```python
count = 0

while count < 5: # prints numbers from 0 to 4
    print(count)
    count += 1
```
