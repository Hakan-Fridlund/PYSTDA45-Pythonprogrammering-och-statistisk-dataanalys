## (Kör först) Skapa veckans dataset

````python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 160

study_hours = rng.uniform(1, 20, size=n)
noise = rng.normal(0, 8, size=n)
exam_score = 25 + 3.0 * study_hours + noise
exam_score = np.clip(exam_score, 0, 100)

df = pd.DataFrame({
    "study_hours": np.round(study_hours, 1),
    "exam_score": np.round(exam_score, 1)
})

missing_rows = rng.choice(df.index, size=4, replace=False)
df.loc[missing_rows, "exam_score"] = np.nan

df.to_csv("study_results.csv", index=False)
print("Skapade study_results.csv med", len(df), "rader.")
````

### Om datasetet (`study_results.csv`)

Det här är ett syntetiskt dataset med studietid och provresultat. Varje rad representerar en studentobservation.

| Kolumn | Typ | Beskrivning |
|---|---|---|
| `study_hours` | decimaltal | Antal studietimmar |
| `exam_score` | decimaltal | Provresultat mellan 0 och 100 |

Datasetet är konstruerat så att det finns ett positivt linjärt samband med slumpmässig variation. Några provresultat saknas för att datakontrollen ska bli en del av analysen.

Eftersom data är syntetiska kan analysen användas för att träna metoder, men den säger ingenting om verkliga studenter eller om orsakssamband mellan studietid och resultat.
