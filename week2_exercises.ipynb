### Uppgift 2 – Strängmanipulation
Läs in en mening och skriv ut den på tre sätt:
- utan vokaler
- utan konsonanter
- med orden i omvänd ordning  

Mellanslag, siffror och skiljetecken behöver inte tas bort. När orden skrivs i omvänd ordning ska bokstäverna i varje ord behålla sin vanliga ordning.

```python
def no_vowels(string):
    vowels = "aeiouåäöAEIOUÅÄÖ"
    result = ""
    for char in string:
        if char not in vowels:
            result += char
    return result

def no_consonants(string):
    consonants = "bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"
    result = ""
    for char in string:
        if char not in consonants:
            result += char
    return result

def reverse_string(string):
    return string[::-1]

input_string = input("Enter a sentence: ").strip()
print(f"Without vowels: {no_vowels(input_string)}")
print(f"Without consonants: {no_consonants(input_string)}")
print(f"Reversed: {reverse_string(input_string)}")
```

### Uppgift 3 – Lista med tal
Låt användaren mata in en lista med tal (separerade med kommatecken, t.ex. `1,4,7,3`).
Programmet ska sedan:
- räkna ut summan
- hitta minsta och största talet
- räkna ut medelvärdet
(Använd loopar, inte `min`, `max`, `sum`).

```python
numbers = [int(x) for x in input("Enter a list of numbers separated by commas: ").strip().split(",")]


def calculate_mean(numbers):
    count = 0
    for i in numbers:
        count += i
    return count / len(numbers) if numbers else 0


def calculate_sum(numbers):
    count = 0
    for i in numbers:
        count += i
    return count

def calculate_min(numbers):
    min_value = numbers[0]
    for i in numbers:
        if i < min_value:
            min_value = i
    return min_value


def calculate_max(numbers):
    max_value = numbers[0]
    for i in numbers:
        if i > max_value:
            max_value = i
    return max_value

print(f"Sum: {calculate_sum(numbers)}")
print(f"Smallest: {calculate_min(numbers)}")
print(f"Largest: {calculate_max(numbers)}")
print(f"Average: {calculate_mean(numbers)}")


```

### Uppgift 5 – Primtal
Skriv en funktion `is_prime(n)` som returnerar `True` om `n` är ett primtal, annars `False`. Ett primtal är ett heltal större än 1 som endast är delbart med 1 och sig självt.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True  # if not divisible on any of the loop iterations, starting from 2 it passes and is returned as a prime

input_number = int(input("Enter a number to check if it's prime: "))
if is_prime(input_number):
    print(f"{input_number} is a prime!")
else:
    print(f"{input_number} is NOT a prime!")
```

### Uppgift 7 – FizzBuzz som funktion
Skriv om din gamla FizzBuzz-kod från vecka 1 till en funktion `fizzbuzz(n)` som returnerar en lista med resultaten.

```python
n = input("Skriv ett heltal:").strip()

def fizzbuzz(n):
    fizzlist = []

    for i in range(int(n)):
        if int(i+1) % 3 == 0 and int(i+1) % 5 == 0:
            fizzlist.append("FizzBuzz")
        elif int(i+1) % 3 == 0:
            fizzlist.append("Fizz")
        elif int(i+1) % 5 == 0:
            fizzlist.append("Buzz")
        elif int(i+1) % 3 == 0 and int(n) % 5 == 0:
            fizzlist.append("FizzBuzz")
        else:
            fizzlist.append(i+1)
    return fizzlist

print(fizzbuzz(n))
```

### Uppgift 11 – Ordräknare
Be användaren skriva en mening. Räkna hur många gånger varje ord förekommer och spara resultatet i en dictionary.  
Skapa därefter ett set med meningens unika ord. Skriv ut både de unika orden och hur många unika ord meningen innehåller.

```python
words = set()
wordcount = []
sentence = input("Skriv en mening här: ").split()

for i in sentence:
    words.add(i)

def word_count(s):
    for i in s:
        found = False
        
        # Leta igenom listan efter ordet
        for d in wordcount:
            if i in d:
                # 1. Hämta det gamla antalet och spara i count
                count = d[i]
                # 2. Öka count med 1
                count = count + 1
                # 3. Spara tillbaka det nya värdet i dictionaryn
                d[i] = count
                
                found = True
                break
        
        # Om ordet inte fanns i listan sedan tidigare
        if not found:
            count = 1
            wordcount.append({i: count})

word_count(sentence)
print("Unika ord (set):", words)
print("Resultat (lista med dicts):", wordcount)
```

### Uppgift 12 – Telefonbok
Skriv ett menyprogram där användaren kan:
- lägga till en kontakt (namn + nummer)
- söka efter ett namn
- visa alla kontakter
- ta bort en kontakt
- avsluta programmet

Kontakterna sparas i en dictionary `{namn: nummer}`.

```python
contacts = {}

def add_contact():
    name = input("\nName:").lower().title()
    nummer = input("Number:")
    contacts[name] = nummer
    print("Contact:", name, " added")

def search_contact():
    search = input("\nSearch for: ").lower().title()
    
    if search in contacts:
        number = contacts[search]
        print(search + " " + number)
    else:
        print("\nNot found")

def remove_contact():
    name = input("\nWhich contact to remove?:").lower().title()
    if name in contacts:
        contacts.pop(name)
        print("\nContact: ", name, "removed")
    else:
        print("\nContact not found")

def show_all():
    print("\n")
    for key in sorted(contacts):
        print(key, contacts[key])



while True:
    print("\n1. Add contact")
    print("2. Search")
    print("3. Remove contact")
    print("4. Show all")
    print("5. Exit")

    choice = input("Choice: ")


    match choice:
        case "1": add_contact()
        case "2": search_contact()
        case "3": remove_contact()
        case "4": show_all()
        case "5":
            break

        case _:
            print("\nFaulty input")
```

### Projekt 1 – Textanalysator
Be användaren skriva in en längre text. Programmet ska analysera:
- antal ord
- antal unika ord
- antal vokaler
- längsta ordet
Resultatet sparas i en dictionary och skrivs ut.

```python
def text_statistics(text):
    words_number = len(text.split())
    words_list = set(text.split())
    longest_word = ""
    shortest_word = None
    characters = 0
    unique_word_count = len(words_list)
    for word in text.split():
        if len(word) > len(longest_word):
            longest_word = word
        if shortest_word is None or len(word) < len(shortest_word):
            shortest_word = word
    for char in text:
        if char != " ":
            characters += 1


    return words_number, longest_word, shortest_word, characters , unique_word_count

sentence = input("Skriv en mening: ")
words_number, longest_word, shortest_word, characters, unique_word_count = text_statistics(sentence)
print(f"Din mening innehåller: {words_number} ord.")
print(f"Det längsta ordet är: {longest_word}")
print(f"Det kortaste ordet är: {shortest_word} ")
print(f"Det är {characters} bokstäver i texten.")
print(f"Det är {unique_word_count} unika ord.")
```
