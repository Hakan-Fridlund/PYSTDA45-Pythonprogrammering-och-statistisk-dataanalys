### **Dictionaries**
key/value (where key is used instead of index)

```python
person = {
    "name": "Håkan",
    "age": 47
        }

print(person)
print(person["name"])

person["city"] = "Malmö"  # add a new key-value pair to the dictionary
print(person)

del person["age"]  # remove the key-value pair with the key "age"
print(person)

print(person.keys())  # print all the keys in the dictionary
print(person.values())  # print all the values in the dictionary
print(person.items())  # print all the key-value pairs in the dictionary

```

```python
for key, value in person.items():
    print(key, ":", value)  # print each key-value pair in the dictionary
```
