### Uppgift 17 – Budget-sammanställning (CSV→JSON)
Skapa en CSV `transactions.csv` med kolumner `date,category,amount` (positivt för intäkt, negativt för kostnad).
Skriv kod som:
- läser transaktionerna,
- summerar **belopp per kategori** i en dictionary,
- skriver ut en **lista** med `{ "category": "...", "total": ... }` sorterad fallande på `total`,
- sparar listan till `budget_summary.json`.

````python
budget_list = []

with open("files/transactions.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    transactions = list(reader)
    unique_categories = set(t["category"] for t in transactions)
    # a set to get unique categories created via list comprehension


    for c in unique_categories:
        value = 0.0
        for t in transactions:
            if t["category"] == c:
                value = value + float(t["amount"])


        category_sum = {"category":c, "total": value}
        budget_list.append(category_sum)
        # lambda function for sorting dictionaries
        budget_list.sort(key=lambda dictionary: dictionary["total"], reverse=True)


with open("files/budget_summary.json", "w", encoding="utf-8") as file:
    json.dump(budget_list, file, indent=4, ensure_ascii=False)
````

Lägg till `__repr__` i `Student` som visar ett tydligt “debug-format”. Kommentera kort när `__str__` vs `__repr__` syns.

````python
class Student:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
        self._courses = []

    def __str__(self): # used when print() function is called with the object, user friendly output
        return f"name: {self.name}, age: {self.age}, courses: {self._courses}"

    def __repr__(self): # used when repr() function is called, developer output: debugger, lists
        return f"Student(name={self.name!r}, age={self.age!r})"

    def add_course(self, course_name: str):
        self._courses.append(course_name)

    @property
    def count_courses(self):
        return len(self._courses)

    def get_courses(self):
        return self._courses.copy()

s1 = Student("Rune", 19)
s2 = Student("Anna", 35)

print(s2)
s1.add_course("matematik")
print(s1)
print(s1.count_courses)
print(s1.get_courses())
repr(s2)
````

### Uppgift 23 – Arv (Inheritance)
Skapa en basklass `Shape` och underklasser som t.ex: `Rectangle(w,h)`, `Circle(r)` och `Triangle(base, height)`.
- Skapa minst 3 underklasser.
- Alla underklasser ska ha en metod `area()` (för cirkel: `3.14159 * r * r`).
- Skapa minst fyra objekt sammanlagt, med en blandning av de olika underklasserna.
- Lägg objekten i en lista och skriv ut klassnamn och area för varje objekt.

````python
class Shape:
    def __init__(self):
        raise NotImplementedError

    def area(self):
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, w: float,h: float):
        super().__init__()
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r: float):
        super().__init__()
        self.r = r

    def area(self):
        return 3.14159 * self.r * self.r

class Triangle(Shape):
    def __init__(self, b: float,h: float):
        super().__init__()
        self.b = b
        self.h = h

    def area(self):
        return (self.b * self.h) / 2


r1 = Rectangle(10,20)
c1 = Circle(10)
t1 = Triangle(10, 20)
t2 = Triangle(50,1000)
shape_list = [r1, c1, t1, t2]

for s in shape_list:
    # .__name__ takes away the module name before to make prettier prints
    print(f"{type(s).__name__} area: {s.area()}")
````

### Uppgift 24 – Bibliotekssystem (OOP)

Bygg ett litet bibliotekssystem med följande klasser:

- `Book(title, author)` representerar en bok.
- `Member(name)` representerar en medlem.
- `Library` hanterar bibliotekets böcker, medlemmar och utlåningar.

`Library` ska innehålla:

- en lista med bibliotekets böcker,
- en lista med registrerade medlemmar,
- en dictionary `loans` som håller reda på vem som har lånat vilken bok.

Låt `loans` använda bokens titel som nyckel och ett `Member`-objekt som värde. Du kan förutsätta att varje boktitel är unik i det här systemet.

Klassen `Library` ska erbjuda metoderna:

- `add_book(book)` – lägger till en bok,
- `add_member(member)` – registrerar en medlem,
- `borrow_book(member, title)` – lånar ut en tillgänglig bok,
- `return_book(member, title)` – lämnar tillbaka en bok,
- `show_loans()` – visar vilka böcker som är utlånade och till vem.

Programmet ska förhindra att:

- en bok som inte finns lånas ut,
- en redan utlånad bok lånas ut igen,
- fel medlem lämnar tillbaka en bok.

Håll all information i minnet. Inga filer behöver användas i denna uppgift.

````python
class Library:
    def __init__(self):
        self.books: list[Book] = []
        self.members: list[Member] = []
        self.borrowed_books = []


    def add_book(self, book):
        self.books.append(book)

    def add_member(self, member):
        self.members.append(member)

    def borrow_book(self, member: "Member", title: str):
        for book in self.books:
            if book.title == title and not book.is_borrowed:
                print(f"{member.name} har nu lånat {book.title}")
                book.borrowed()
                self.borrowed_books.append({"name": member.name, "book": title})
                return # cuts the loop when found
        print("Boken är tyvärr inte tillgänglig.")


    def return_book(self, member: "Member", title: str):

        for loan in self.borrowed_books:
            if loan["name"] == member.name and loan["book"] == title:
                # finds the book and restores it
                for book in self.books:
                    if book.title == title:
                        book.returned()
                        print(f"{book.title} är nu återlämnad av {member.name}")
                        break
                # Removes the loan from borrowed_books list
                self.borrowed_books.remove(loan)
                return

    def show_loans(self):
        print(f"Utlånade böcker: {self.borrowed_books}")

class Book:
    def __init__(self, title: str, author: str, is_borrowed: bool = False):
        self.title = title
        self.author = author
        self.is_borrowed = is_borrowed

    def borrowed(self):
        self.borrowed = True

    def returned(self):
        self.borrowed = False

class Member:
    def __init__(self, name: str):
        self.name = name

# test code below
stadsbiblioteket = Library()
bok1 = Book("Python testing", "Kalle Karlsson")
medlem1 = Member("Anna")

stadsbiblioteket.add_book(bok1)
stadsbiblioteket.add_member(medlem1)

stadsbiblioteket.borrow_book(medlem1, "Python testing")
stadsbiblioteket.show_loans()

stadsbiblioteket.return_book(medlem1, "Python testing")
stadsbiblioteket.show_loans()
````

### Projekt 2 – Ordfrekvens-pipeline (fil → dict → JSON)
- Läs en textfil, normalisera (små bokstäver, ta bort `, . ! ?`), räkna ord i en dictionary.
- Skriv ut **topp 10** mest frekventa orden.
- Spara hela frekvens-dictionaryn till `word_freq.json`.
- (Valfritt) Läs tillbaka JSON och skriv ut ord som förekommer **exakt** `N` gånger (från input).

````python
import json

def normalize_text(line):
    new_line = []
    for char in line:
        char = char.lower()
        if char in ",.!?/\\":
            continue
        else:
            new_line.append(char)
    return "".join(new_line)

def count_words(lines):
    """returns a dict {word:count} for a list of lines"""
    word_freq = {}
    for line in lines:
        for word in line.split():
            if word in word_freq:
                word_freq[word] += 1
            else:
                word_freq[word] = 1
    return word_freq


with open("files/novell.txt", "r", encoding="utf-8")as file:
    lines = file.readlines() # list with strings for each line

new_lines = []
# appends the normalized lines to new_lines
for line in lines:
    new_lines.append(normalize_text(line))

word_freq = count_words(new_lines)

with open("files/word_freq.json", "w", encoding="utf-8") as file: # saves the word_freq
    json.dump(word_freq, file, ensure_ascii=False, indent=4)


with open("files/word_freq.json", "r", encoding="utf-8")as file: # loads the word_freq
    word_freq = json.load(file)

# lamda function for key sorting, word_freq.items() gives a list with pairs of tuples (word, count)
# ex[("a", 2),("b", 5)], sorted goes through pairs one at a time, the parameter is the pair ("a",2)
# pair[1] means index 1 is the 2 in ("a",2). revers=True sorts the highest first
# [:10] takes indexes 0-9, 10st
top_10 = sorted(word_freq.items(), key=lambda pair: pair[1], reverse=True)[:10]

for rank, (word, count) in enumerate(top_10, start=1):
    print(f"{rank}. {word}: {count}")
````
