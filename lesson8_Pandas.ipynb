# Pandas and dataframes
- **series** is a named and indexed column with data
- **dataframe** is a table with named columns with data, each column can has its own datatype

```python
import pandas as pd

s = pd.Series([10, 20, 30], name="studietimmar")
df_demo = pd.DataFrame({
    "student": ["A", "B", "C"],
    "studietimmar": [10, 5, 8],
    "provpoäng": [85, 70, 78]
})

print(s)
print(df_demo)
```

## pd.read


```python
df = pd.read_csv("files/studietimmar.csv")
print(df.head())
print(df.shape)
print(df.dtypes)
```

```python
df.columns = (df.columns    # this only affect the column names
              .str.strip() #strips spaces from columns
              .str.replace(" ", "_") # replaces space with _
              .str.lower() # lower case
              )
print(df.columns)
```

```python
df.info() # shows missing values and datatypes
```

```python
df.describe() # show statistical keyvalues for numerical columns only
```

```python
df["klass"].value_counts(dropna=False)
```

```python
df[["studietimmar", "provpoäng"]].head()
mask = (df["studietimmar"] >= 5) & (df["provpoäng"] >= 70)
 # mask to filter all studietimmar below 5 and score below 70 | is used for OR
 # ~ negates the whole expression
df[mask].head() 
```

- loc  
rows index, column names  
- iloc  
rows index, column index

```python
print(df.loc[0:3, ["student", "studietimmar"]])
print(df.iloc[0:3, 0:2])
```

```python
df.sort_values("provpoäng", ascending=False).head(3)
```

```python
df_renamed = df.rename(columns={"provpoäng": "poäng"}) # rename a column
df_renamed.head()
```
