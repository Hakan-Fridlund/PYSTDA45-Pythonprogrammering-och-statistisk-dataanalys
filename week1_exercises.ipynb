### 4. Strängar – indexering, slicing, sammanslagning
Utgå från `name = "Ada Lovelace"`.
1. Skriv ut antalet tecken, de tre första tecknen och de två sista tecknen i strängen.
2. Plocka ut förnamnet och efternamnet (hint: `split`).
3. Skriv `Hej, <förnamn>!` med f-string.
4. Skapa en ny sträng där mellanslag ersätts med `_`.
5. Skriv ut strängen baklänges med slicing.

```python
name = "Ada Lovelace"

print(f"Number of characters in the name: {len(name)}")
print(name[0:3])
print(name[-2:])
first_name, last_name = name.split()  #unpacking the first and last name into separate variables
print(f"Hej, {first_name}!")

sentence = "Skapa en ny sträng där mellanslag ersätts med_"
new_sentence = sentence.replace(" ", "_")
print(new_sentence)
print(new_sentence[::-1]) # prints the new string in reverse order [start:stop:step]
```

### 9. Listor – redigera & summera

utgå ifrån listan nums i den förra uppgiften och skriv kod som:

1. Byter ut det **andra** elementet i listan mot `100`.

4. Tar fram **minsta** och **största** tal **utan** att använda `min`/`max` – använd egna variabler och jämförelser i en loop.

```python
nums[1] = 100
print(nums)
count = 0

for i in nums:
    count = count + i
print(count)

print(sum(nums))

smallest_number = 1000
for i in nums:
    if i < smallest_number:
        smallest_number = i
print(smallest_number)

biggest_number = 0
for i in nums:
    if i > biggest_number:
        biggest_number = i
print(biggest_number)

```

### 10. For-loopar

2. Skriv siffrorna 1–20 på en rad, men ersätt jämna tal med `*`.
3. Skapa en lista med kvadraterna av talen 1–10.

```python
words = ["python", "data", "loop"]
for i in words:
    print(f"{i} is {len(i)} letters long.")


for i in range(1, 21):
    if i % 2 == 0:
        print("*", end=" ")
    else:
        print(i, end=" ")

squared_numbers = []
for i in range(1, 11):
    squared_numbers.append(i**2)

print(f"\n{squared_numbers}")
```
