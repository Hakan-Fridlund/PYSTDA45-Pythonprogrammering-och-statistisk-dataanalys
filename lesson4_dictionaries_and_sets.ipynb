### **Dictionaries**
key/value (where key is used instead of index)

```python
person = {
    "name": "Håkan",
    "age": 47
        }

print(person)
print(person["name"])

person["city"] = "Malmö"  # add a new key-value pair to the dictionary
print(person)

del person["age"]  # remove the key-value pair with the key "age"
print(person)

print(person.keys())  # print all the keys in the dictionary
print(person.values())  # print all the values in the dictionary
print(person.items())  # print all the key-value pairs in the dictionary

```

```python
for key, value in person.items():
    print(key, ":", value)  # print each key-value pair in the dictionary
```

### **Sets**
Does NOT contain duplicates and is not ordered

```python
fruits = {"apple", "banana", "orange", "apple"}
print(fruits)
```

### **Tuples**
Good for constants like coordinates, not allowed to change variables in a tuple

```python
coord_tuple = (3, 4)
# coord_tuple[0] = 10 # this code will get an error message when trying to change

print(coord_tuple[0])
```

### Unpacking

```python
p = (3, 4)

x, y = p # unpacking variables

print(x)
print(y)
```

#### Tuple common functions

```python
t = (10, 20, 30, 40, 50)

print(t[1:4])

a = (1, 2)
b = (3,2)
c = a + b
print(c)

print(c.count(2)) # prints the number of 2s in the tuple
print(c.index(3)) # prints the indexnumber that contains the 3 in the tuple

head, *rest = (100, 200, 300, 400) # unpacking with * that will add the rest from position
print(head)
print(rest)
```
